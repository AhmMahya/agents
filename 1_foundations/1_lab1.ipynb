```python
from openai import OpenAI
from dotenv import load_dotenv
import os


load_dotenv(override=True)
gemini_api_key = os.getenv("GOOGLE_API_KEY")
gemini_base_url = "https://generativelanguage.googleapis.com/v1beta/openai/"

gemini = OpenAI(api_key=gemini_api_key, base_url=gemini_base_url)
```

```python
query_1 = "peack a business area that might be worth exploring for an Agentic AI"

messages_1 = [
    {
        "role": "user",
        "content": query_1
    }
]

response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=messages_1)
business_area= response.choices[0].message.content
```

```python
from IPython.display import Markdown, display
display(Markdown(business_area))
```

```python
query_2 = f"""
based on the business area in below:
{business_area}

present a pain-point in that industry, something challenging that mught be ripe for an Agentic solution.
"""

messages_2 = [
    {
        "role": "user",
        "content":query_2
    }
]

response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=messages_2)
pain_point = response.choices[0].message.content
```

```python
display(Markdown(pain_point))
```

```python
query_3 = f"""
based on the pain-point in below:
{pain_point}

for the business area in below:
{business_area}

give an Agentic AI solution for that pain-point
"""

message_3 = [
    {
        "role": "user",
        "content": query_3
    }
]

response = gemini.chat.completions.create(model="gemini-2.5-flash-lite", messages=message_3)
Solution = response.choices[0].message.content
```

```python
display(Markdown(Solution))
```
